# Run the Zirconia O2 app
Starts `app.py` (Flask) as a background process. Open http://127.0.0.1:5050. Run the last cell to stop it.

```python
import subprocess, sys, time, os, urllib.request
os.chdir(os.path.abspath('.'))
proc = subprocess.Popen([sys.executable, 'app.py'], stdout=open('app.log', 'w'), stderr=subprocess.STDOUT)
for _ in range(60):
    try:
        urllib.request.urlopen('http://127.0.0.1:5050/api/results', timeout=1); break
    except Exception:
        time.sleep(0.5)
print(open('app.log').read())
print('UI running at http://127.0.0.1:5050  (pid', proc.pid, ')')
```

```python
# Optional: quick API check
import json
print(json.load(urllib.request.urlopen('http://127.0.0.1:5050/api/predict?P=56.89&I=8.082'))['C_corr'])
```

```python
# Stop the server
proc.terminate(); proc.wait(); print('stopped')
```
